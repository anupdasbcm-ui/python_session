```python
# WAP ask the user get one random number between 1 to 20
# enter a number
# compare these two number
# if both are same: print("you won")
# else:print(lost)

#M-1: Write in normal way
#M-2: Create a function without argument
#M-3: Create a function with argument
```

```python
#M-1:
import random
random_num=random.randint(1,20)
num=eval(input("enter a number:"))

if random_num==num:print("you won")
else:print("you lost")
```

```python
#M-2
import random
def compare():
    random_num=random.randint(1,20)
    num=eval(input("enter a number"))
    
    if random_num==num:
        print('you won')
    else:
        print('you lost because the random number is:',random_num)
compare()
    
```

```python
#M-3
import random
def compare(num):
    random_num=random.randint(1,20)
    
    if random_num==num:
        print('you won')
    else:
        print('you lost because the random number is:',random_num)
compare(9)
```

```python
 import random
def compare(random_num,num):

    if random_num==num:
        print("you won")
    else:
        print("you lost because of the random number is:",random_num)

value_random=random.randint(1,20)
value=eval(input("enter number"))
compare(value_random,value)
```

```python
def avg(num1,num2,num3=40):
    add=(num1+num2+num3)
    print("num1:",num1)
    print("num2:",num2)
    print("num3:",num3)
    num3=80
    add=(num1+num2+num3)
    avg=add/3
    print("The average is:",avg)

avg(20,30,60)

# while defining function: num3=40
# while you are calling function num3=80
# after enter inside the function num3 =60
```
