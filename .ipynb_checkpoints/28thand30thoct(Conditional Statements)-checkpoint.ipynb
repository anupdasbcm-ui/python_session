```python
num1 = eval(input("Enter a number:"))
if(num1>0):
    print("the given number {} is positive.".format(num1))
elif num1<0:
    print("the given number {} is negative".format(num1))
else:
    print("the given number {} is zero.".format(num1))
```

```python
# wap ask the user to enter percentage of marks
# if %tage of marks> 90, print A grade
# if %tage of marks between 70 to 90, print B grade
# if %tage of marks between 50 to 70, print C grade
# ohterwise print D
# if-elif-elif-else
```

```python
per=eval(input("enter your percentage: "))
if per>90:
    print ("A grade")
elif 70<per>90:
    print ("B grade")
elif 50<per>70:
    print ("C grade")
else:
    print("D grade")
```

```python
# wap ask the use enter distance in km
# ask the user enter rate per km
# if distance is more than 10 km, print charge is 100rs
# if distance is between 7 to 10 km, print charge is 70 rs
# if distance is between 4to 7 km, then print charge is 40rs
# otherwise, print charge is 30rs
```

```python
Distance=eval(input("Enter a distance in Km:"))
if(Distance>=10):
    print("The charge is 100rs")
elif(7<=Distance<10):
    print("The charge is 70rs")
elif(4<=Distance<7):
    print("The charge is 40rs")
else:
    print("The charge is 30rs")
```

```python
# wap ask the user enter salary 
# if salary > 10 lacs: enter the user tax percentage: 10%, print the amt of tax paid :

# if  salary between 7 lacs to 10 lacs, enter the user tax percentage: 5%, print the amount of tax paid.
#if salary between 4 to 7 lakhs, enter the user tax percentage:2.5%, print the amount of tax paid 
# otherwise: print ('No tax required')
```

```python
Salary=eval(input("Enter the salary:"))
if(Salary>=1000000):
    print("The tax is 10% and amt of tax paid is {}".format(Salary*0.1))
elif(700000<=Salary<1000000):
    print("The tax is 5% and amt of tax paid is {}".format(Salary*0.05))
elif(400000<=Salary<700000):
    print("The tax is 2.5% and amt of tax paid is {}".format(Salary*0.025))
else:
    print("No tax required")
```

```python
# Wap ask the user enter which operation you need to perform
# Operations are : addition/multiplication/subtraction/division
# ask the user enter a number1
# ask the user enter a number2
# if the operation equal to addition
#         print("we are doing addition operation")
#         print("The addition of num1 and num2 is ")
# if the operation equal to multiplication
#         print("we are doing multiplication operation")
#         print("The multiplication of num1 and num2 is ")
# if the operation equal to subtraction
#         print("we are doing subtraction operation")
#         print("The subtraction of num1 and num2 is ")
# if the operation equal to division
#         print("we are doing division operation")
#         print("The division of num1 and num2 is ")
# else:
#     print("provide suitable operation")
```

```python
num1=eval(input("Enter number1:"))
num2=eval(input("Enter number2:"))
Operation=input("Enter operation:")
if Operation=="addition":
    print(" we are doing addition operation")
    print("The addition of num1 and num2 is {}".format(num1+num2))

elif Operation=="multiplication":
    print(" we are doing multiplication operation")
    print("The multiplication of num1 and num2 is {}".format(num1*num2))

elif Operation=="subtraction":
    print(" we are doing subtraction operation")
    print("The subtraction of num1 and num2 is {}".format(num1-num2))

elif Operation=="division":
    print(" we are doing division operation")
    print("The division of num1 and num2 is {}".format(num1/num2))

else:
    print("Provide suitable operation")

    
```

```python
try:
    num1=eval(input("Enter number1:"))
    num2=eval(input("Enter number2:"))
    Operation=input("Enter operation:")
    if Operation=="addition":
        print(" we are doing addition operation")
        print("The addition of num1 and num2 is {}".format(num1+num2))

    elif Operation=="multiplication":
        print(" we are doing multiplication operation")
        print("The multiplication of num1 and num2 is {}".format(num1*num2))

    elif Operation=="subtraction":
        print(" we are doing subtraction operation")
        print("The subtraction of num1 and num2 is {}".format(num1-num2))

    elif Operation=="division":
        print(" we are doing division operation")
        print("The division of num1 and num2 is {}".format(num1/num2))

    else:
        print("Provide suitable operation")

except Exception as e:
    print(e)

    
```

```python
try:
    num1=eval(input("Enter number1:"))
    num2=eval(input("Enter number2:"))
    Operation=input("Enter operation:")
    if Operation=="1":
        print(" we are doing addition operation")
        print("The addition of num1 and num2 is {}".format(num1+num2))

    elif Operation=="2":
        print(" we are doing multiplication operation")
        print("The multiplication of num1 and num2 is {}".format(num1*num2))

    elif Operation=="3":
        print(" we are doing subtraction operation")
        print("The subtraction of num1 and num2 is {}".format(num1-num2))

    elif Operation=="4":
        print(" we are doing division operation")
        print("The division of num1 and num2 is {}".format(num1/num2))

    else:
        print("Provide suitable operation")

except Exception as e:
    print(e)
```
