```python
string1='ola ola ola'
# ola   ola   ola
# 012 3 456 7 8910
# we are counting the number of 'a' from index 4
string1.count('a',4)
print(string1.count('a',4))
print(string1.count('a',6))
print(string1.count('a',4,6))  #4 and 5
print(string1.count('a',4,7))  #4 5 6
print(string1.count('A',4))
print(string1.count('A'.lower(),4))
```
