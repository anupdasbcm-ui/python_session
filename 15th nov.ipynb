$strip$-$lstrip$-$rstrip$

```python
str1=' hello how are you '
str2=" hello how are you"
str3="hello how are you "

# I want to remove the spaces
# if you want to remove spaces both side use strip method
# if you want to remove the spaces only left side then use lstrip: left strip
# if you want to remove the spaces only right side then use rstrip: right strip
```

```python
print(str1.strip())
print(str1.lstrip())
print(str1.rstrip())
```

```python
print(str1.strip())
print(str2.lstrip())
print(str3.rstrip())
```

```python
str1='python.anaconda@anupit.com'
str1.index('.')

second_dot=str1.index('.',6+1)  # after completion of first dot
```
