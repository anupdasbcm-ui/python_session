```python
# Wap create a dictionary based on two lists.
#list1=['Ram','Raheem','Robert']
#age=[25,30,35]
list1=['Ram','Raheem','Robert']
age=[25,30,35]
dict1={}
for i,j in zip(list1,age):
    dict1[i]=j

dict1
```

```python
dict1={}
for i in range(len(list1)):
    dict1[list1[i]]=age[i]

dict1
    
```

```python
# dictionary comprehension.
dict1={}
dict1={list1:age for list1,age in zip(list1,age)}


dict1
```

```python
dict1={}
dict1={list1[i]:age[i] for i in range(len(list1))}

dict1
```

```python
# wap take 5 random no. and create a dictionary with even and odd
# step-1 : take empty dictionary
# step-2 : take two even and odd list
# step-3: import random
# step-4: for i range (5):
# step5: num=random.randint(a,b)
# step-6: if <even>:
# step-7: append the values in even list
# step-8: else:
# step-9: append the values in odd list
# step-10: create dictionary

dict2={}
list1=[]
list2=[]
import random
for i in range(5):
    num=random.randint(1,100)
    if num%2==0:
        list1.append(num)
    else:
        list2.append(num)

print('list1:',list1)
print('list2:',list2)
dict2={'even':(list1),'odd':(list2)}
dict2
```
