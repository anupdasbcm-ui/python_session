```python
# with argument
def sum_avg(n1,n2,n3):
    summ=n1+n2+n3
    avg=summ/3
    print('sum:',summ)
    print('avg',avg)
    return(summ,avg)

addition,average=sum_avg(20,30,40)

# sum_avg(20,30,40) is returning two values
# summ and avg
# summ we are saving in a variable: addition
# avg we are saving in a variable: average
print(addition,average)
```

```python
# Find the biggest number of 3 numbers
# take three numbers
# and find the biggest number
# and return that number

#num1>num2 and num1>num3
```

```python
def biggest_num(n1,n2,n3):
    
    if n1>n2 and n1>n3:
        print("n1 is bigger")
        return(n1)
    elif n2>n3:
        print("n2 is bigger")
        return(n2)
    else:
        print("n3 is bigger")
        return(n3)
Biggest=biggest_num(20,200,10)


#it will return only one value, whenever condition is satisfied.

print(Biggest)
```

```python
def add(a,b):
    return(a+b)

# mul
def mul(a,b):
    return(a*b)

# sub
def sub(a,b):
    return(a-b)

# div
def div(a,b):
    return(a/b)

val_add=add(20,30)
val_sub=sub(20,30)
val_mul=mul(20,30)
val_div=div(20,30)
print(val_add,val_sub,val_mul,round(val_div,2))
```

```python
def aggregation(a,b):
    return(a+b,a-b,a*b,a/b)

add,sub,mul,div=aggregation(20,30)
print(add,sub,mul,round(div,2))
```
