***Local Variables***

```python
# Local variables means the variables provided inside the function.

#Global variables means the variables provided outside the function.
```

```python
# Local variables example.

def addition():
    n1=10
    n2=20
    add=n1+n2
    print("the addition of {} and {} is {}".format(n1,n2,add))
```

```python
# Global variables example.

n1=20
n2=30
def addition():      # No need to provide inside brackets
    #no need to mention
    add=n1+n2
    print("The addition of {} and {} is {}".format(n1,n2,add))

addition()
```

```python
def avg():
    
    num11=20
    num22=30
    num33=40
    avg=(num11+num22+num33)/3
    print("the avg of {},{} and {} is {}".format(num11,num22,num33,avg))
    return(num11,num22)

val1,val2=avg()
```

```python
n1=20
n2=30
def addition(n1=700):
    n2=300
    add=n1+n2
    print("the addition of {} and {} is {}".format(n1,n2,add))

addition(500)

print(n1)

# step-1: n1=20
# step-2: n2=30
# step-3: function defined n1=700
# step-4: calling the function:n1=500 ======>
# step-5:                      n2=300
# step-6:                         800
# step-7:n1=20
```

```python
def value():
    global number1,number2
    number1=10
    number2=20
    print(number1)

value()
print(number1)
print(number2)
```
