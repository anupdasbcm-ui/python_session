$pattern-3$

```python
for i in range(start,stop)

- starr: start of the loop, initial point

- direction: what is sign of step value, that is the direction
            - direction will not decide by start and stop value

            - direction will provided by step value

- if step size is positive direction
    
          end=stop-1

- if step is negative direction

          end=stop+1
```

```python
for i in range(2,20,2):
    print(i,end=' ')

# start=2
#direction: step size=+2
#end=stop-1=20-1=19
```

```python
for i in range(-1,-10,-1):
    print(i,end=' ')

# start= -1
# direction: step= -1     negative
# end= stop+1 = -10+1= -9
# is this possible or not?
```

```python
for i in range(-1,-10,1):
    print(i,end=' ')

# start=-1
# direc= step=+1
# end=stop-1= -10-1= -11
# possible/np
```

```python
for i in range(8,20,-2):
    print(i)

#start=8
#direc=step=-2  :neg
# end=stop+1=20+1=21
```
