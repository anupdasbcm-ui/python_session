```python
# Wap ask the user get 10 random numbers b/w 0 to 10.
# count how many are greater than 5 or equal=0
```

```python
count=0
i=0
while i<10:
    num=random.randint(0,10)
    if num>5 or num==0:
        print(num,end=' ')
        count+=1

    i=i+1
print('/')
print(count)
```
