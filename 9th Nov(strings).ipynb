```python
# wap ask the user find number of 'a' letters in a given string
# string="Hai how are you and how do you do"
# ans:3

#first iterate the string using in operator and for loop
# print each letter ======>  i
# apply the if condition i== 'a' 
```

```python
count =0
string='Hai how are you and how do you do'
for i in string:
    
    if i=='a':
        count=count+1

print(count)

# step-1: i='h'  'h'=='a' F
# step-2: i='a'  'a'=='a'  T === > count=1
        
```
